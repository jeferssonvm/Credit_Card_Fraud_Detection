```python

corr = df.corr()

# 2. Generamos una máscara para el triángulo superior
mask = np.triu(np.ones_like(corr, dtype=bool))

# 3. Configuramos el gráfico
plt.figure(figsize=(16, 14)) 

# 4. Dibujamos el heatmap con la máscara
sns.heatmap(corr, 
            mask=mask,         
            annot=True,         
            fmt=".2f",          
            cmap='coolwarm',   
            vmin=-1, vmax=1,    
            linewidths=0.5)     

plt.title("Matriz de Correlación Limpia (Triángulo Inferior)")
plt.show()
```

```python
df_numeric = df.select_dtypes(include=['number']).columns
for col in df_numeric:
    plt.figure(figsize=(10, 6))
    sns.histplot(df[col], bins=50, kde=True)
    plt.title(f'Distribución de {col}')
    plt.xlabel(col)
    plt.ylabel('Frecuencia')
    plt.show()
```

```python
for col in ['V17', 'V14', 'V12', 'V11']: # Empieza por las más correlacionadas
    plt.figure(figsize=(8, 4))
    sns.boxplot(data=df, x='Class', y=col)
    plt.title(f'¿Cómo varía {col} según la Clase?')
    plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Class', data=df)
plt.title('Conteo de Clases (0: Normal, 1: Fraude)')
plt.yscale('log') # Escala logarítmica para poder ver la barra del fraude
plt.show()
```

```python
import sweetviz as sv

# Comparar las distribuciones basadas en la variable Class
# Opción: Solo variables con alta correlación + Amount y Time
cols_interes = ['V17', 'V14', 'V12', 'V10', 'V11', 'V4', 'Amount', 'Time', 'Class']

my_report = sv.compare_intra(df[cols_interes], df["Class"] == 1, ["Fraude", "Normal"])
my_report.show_html("reporte_comparativo.html")
```
